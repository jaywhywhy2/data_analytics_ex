- axis = 0 : 행 인덱스가 증가하는 방향
- axis = 1 : 열 인덱스가 증가하는 방향

```python
# 각 열별 최대 값을 구하고 싶다.
# [94, 40, 97, 73, 82]
# 가장 큰 값을 구해야 하는 경우

X.max(axis=0)
```

```python
# 각 행별 최대 값을 구하고 싶다.
# [73, 75, 97, 94]
# 가장 큰 값을 구해야 하는 경우

X.max(axis=1)
```
