```python
import folium
```

```python
#디폴트 위도/경도 지점(focusing), 지도 사이트인 map객체를 생성
folium.Map(location=[37.5662952, 126.9779451], zoom_start=16)
```

```python
folium_map = folium.Map(location=[37.5662952, 126.9779451], zoom_start=16)
folium.Marker(location=[37.5662952, 126.9779451]).add_to(folium_map) #메모리에 씌우기
```

```python
import requests
import folium
# from pandas.io.json import json_normalize <-- 아래와 같이 수정
from pandas import json_normalize
```

```python
# 스벅 위치 데이터 가져오기
targetSite='https://www.starbucks.co.kr/store/getStore.do?r=8K3XF06R9Q'
req=requests.post(targetSite, data={
'ins_lat': 37.563398,    # 위도
'ins_lng': 126.9863309,  # 경도
'p_sido_cd': '01',
'p_gugun_cd': '',
'in_biz_cd': '',
'iend': 300,   # 300개
'set_date': ''
})
```

```python
starbucks=req.json()
```

```python
print(type(starbucks))
print(starbucks.keys())
```

**[tip] json_normalize(starbucks, 'list') 의미**
- 'list' 키의 데이터를 df로 만들어줌

```python
data= {
    "status": "success",
    "count": 5,
    "list": [
        {"store_id": "001", "name": "Store A", "location": "Seoul"},
        {"store_id": "002", "name": "Store B", "location": "Busan"}
    ]
}
data
```

```python
result = json_normalize(data, 'list')
result
```

```python
starbucks_df=json_normalize(starbucks, 'list')
starbucks_df
```

```python
# df 정보 보기
# starbucks_df.info()
starbucks_df.info(verbose=True, memory_usage='deep', show_counts=True)
```

```python
starbucks_df.columns
```

```python
cols = ['s_name','sido_code','sido_name','gugun_code','gugun_name', 'doro_address','lat','lot']
starbucks_df_map=starbucks_df[cols]
starbucks_df_map.head(3)
```

```python
# 위도, 경도 숫자로 변경
starbucks_df_map.loc[:, 'lat']=starbucks_df_map['lat'].astype(float)
starbucks_df_map.loc[:, 'lot']=starbucks_df_map['lot'].astype(float)
```

```python
starbucks_df_map.head(3)
```

```python
# 지도에 표현하기 일단 1개 실습
# step1. 일단 데이터에 starbucksMap이란 걸 넣어놓기
starbucksMap=folium.Map(location=[37.501087,127.043069], zoom_start=16)
# step2. 팝업 설정
popup=folium.Popup('역삼아레나빌딩점', max_width=300)
euksam_location = [starbucks_df_map[starbucks_df_map['s_name'] == '역삼아레나빌딩']['lat'].values[0],
                   starbucks_df_map[starbucks_df_map['s_name'] == '역삼아레나빌딩']['lot'].values[0]]
folium.Marker(location= euksam_location, popup=popup).add_to(starbucksMap)
starbucksMap
```

```python
popup=folium.Popup('종각점', max_width=300)
```

```python
folium.Marker(location=[37.569918, 126.984528], popup=popup).add_to(starbucksMap)
popup=folium.Popup('종로관철점', max_width=300)
folium.Marker(location=[37.569058, 126.986013], popup=popup).add_to(starbucksMap)
popup=folium.Popup('종로구청점', max_width=300)
folium.Marker(location=[37.572662, 126.981138], popup=popup).add_to(starbucksMap)
```

# [문제] 위에 있는 모든 지역을 지도에 표현하기

```python
starbucks_df_map.head(3)
```

-  s_name, lat, lot 컬럼 추출

```python
starbucks_map2 = starbucks_df_map[['s_name','lat','lot']]
starbucks_map2.head()
```

- 데이터 프레임의 values만 추출

```python
s_map_datas = starbucks_map2.values
s_map_datas[:5]
```

### 지도에 표현하기

```python
s_map_datas[0][1]
```

```python
len(s_map_datas)
```

```python
# 지도 객체 생성
foliumMap=folium.Map(location=[s_map_datas[0][1], s_map_datas[0][2]], zoom_start=16)

#저장할 폴더 생성
os.makedirs('folium_data', exist_ok=True)

# map에 표시하기
n=0
for n in range(len(s_map_datas)):
    popup=folium.Popup(s_map_datas[n][0], max_width=300)
    folium.Marker(location=[s_map_datas[n][1],s_map_datas[n][2]], popup=popup).add_to(foliumMap)
    
# 지도 파일로 저장
foliumMap.save('folium_data/folium_map1.html')
# foliumMap 객체 반환
foliumMap
```
