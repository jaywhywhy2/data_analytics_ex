## 실습 10. '외국인비율' 과 '고령자비율' 두개의 컬럼을 추가하세요.
- 외국인비율 = 외국인 수 / 인구수 * 100
- 고령자비율 = 고령자 수 / 인구수 * 100

```python
pop_Seoul.head()
```

```python
# 외국인비율 구해서 df에 적용하기
pop_Seoul['외국인비율'] = pop_Seoul['외국인'] / pop_Seoul['인구수']*100
pop_Seoul.head()
```

```python
# 고령자비율 구해서 df에 적용하기
pop_Seoul['고령자비율'] = pop_Seoul['고령자'] / pop_Seoul['인구수']*100
pop_Seoul.head()
```

## 실습 16. 상관 관계 분석에 대해서 알아봅니다.
### https://ko.wikipedia.org/wiki/%EC%83%81%EA%B4%80_%EB%B6%84%EC%84%9D

상관계수는 -1부터 1까지의 값을 갖는다.

1일 때 완벽한 양의 상관관계가 되고, 

-1일 때 완벽한 음의 상관관계가 된다. 

0이라면 별다른 상관관계가 없음을 의미한다.


-  인공지능을 만들때 관련이 있는 데이터를 추출하기 위함. -> 정확도가 중요함.

## 실습 20. 각 구의 CCTV의 갯수를 bar로 나타내세요.
### pandas dataframe.plot 함수 사용

```python
#이 실습파일에서는 위에서 깔았으니 안해도 되지만, 기억차원에서 ㅎㅎ
import pandas as pd
import matplotlib.pyplot as plt
```

- 선 그래프 (기본값)<br>
df.plot(kind='line')<br>
df.plot.line()

- 막대 그래프<br>
df.plot(kind='bar')<br>
df.plot.bar()

- 수평 막대<br>
df.plot(kind='barh')<br>
df.plot.barh()

- 히스토그램<br>
df.plot(kind='hist')<br>
df.plot.hist()

- 박스플롯<br>
df.plot(kind='box')<br>
df.plot.box()

- 산점도<br>
df.plot(kind='scatter', x='컬럼1', y='컬럼2')<br>
df.plot.scatter(x='컬럼1', y='컬럼2')

- 영역 그래프<br>
df.plot(kind='area')<br>
df.plot.area()

- 파이 차트v
df.plot(kind='pie', y='컬럼')<br>
df.plot.pie(y='컬럼')

- 밀도 그래프<br>
df.plot(kind='density')<br>
df.plot.density()

- 육각형 밀도<br>
df.plot(kind='hexbin', x='컬럼1', y='컬럼2')<br>
df.plot.hexbin(x='컬럼1', y='컬럼2')

**기본 옵션**
- kind : 그래프의 종류
- x=, y= : 축 칼럼
- title='' : 제목
- figsize=(10, 6) : 그래프 크기 (가로, 세로)
- grid=True : 격자 표시
- legend=True : 범례 표시
- xlabel='',ylabel='' : x축, y축 라벨
- color='blue' : 색상
- style='--' : 선 스타일
- alpha=0.7 :투명도 (0~1)
- rot=45 : x축 라벨 회전
- fontsize=12 : 폰트 크기

```python
# x 축은 인덱스, y축은 데이터로 표시하는 함수
data_result.plot()
plt.show()
```

```python
data_result.head(2)
```

```python
# CCTV갯수 시각화
data_result.plot( y='CCTV갯수',kind='line')
plt.show()

#오히려 x축 지정 안해야 하는 상황 ('구별'을 index로 사용하고 있어서)
```
