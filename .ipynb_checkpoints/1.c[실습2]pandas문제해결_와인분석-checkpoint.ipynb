```python
# 로컬에서 로딩

data = "./datas/winemag-data.csv"

# 첫번째 컬럼을 index에 사용 #index_col 파라미터의 의미는, 지정한 것을 index칼럼으로 쓰겠다는 뜻!
#index_col=1을 하면 해당 예시에서는, country를 index 칼럼값으로 사용!
df = pd.read_csv(data, index_col=1)
df.head(2)
```

```python
df = pd.read_csv(data, index_col=0)
df.head(2)
```

```python
# 1. title 컬럼에 데이터가 비어있는것이 있는지 확인한다.
df.info()
```

```python
df['title'].isna() #비어있는 값 있냐? 확인
```

df['title'] → title 컬럼 선택
- isna() → 결측치면 True, 아니면 False
- sum() → True를 1로 계산해서 합산
- 값이 0이 나오면, 비어있는 값이 없는 것!

```python
df['title'].isna().sum() 
```

```python
#각 셀이 결측치가 아닌지(not NA) 확인하여 True/False로 반환
#결측치 : NaN, None, NaT 등 값이 없는 것
df.notna() 
```

- df.notna() → True/False 데이터프레임 생성
- sum() → True를 1로, False를 0으로 계산해서 합산
- 단, 각 column에 따라서 보여줌 > 즉 칼럼마다 이 값이 다르면 결측값이 존재!

```python
df.notna().sum() #결측치가 아닌 값(유효한 데이터)의 개수
```

```python
df.columns
```

```python
df['title'].unique()
```

```python
df['title'].nunique() #title 칼럼에서 고유한 값(유효값)의 갯수를 반환!
```

##  문제2

```python
# 각 테이스터들은, 각각 몇개의 와인을 테스트 했는지 확인하시오. ( 테스터 이름, 갯수 )
# 포인트의 평균을 구하고, 리뷰의 포인트값이, 평균보다 큰 데이터 (즉, 평가가 좋은 와인) 만 가져오시오.
# points 컬럼의 median 값은?
# 각 국가별로는 몇개의 리뷰가 있는지, 각국가별 리뷰수를 구하시오.
# 나는 경제적이므로, 가격대비 포인트가 가장 큰 와인을 사려한다. 해당 와인의 title은?
```

```python
df = pd.read_csv(data, index_col=0)
df.head(2)
```

```python
df['taster_name'].value_counts()
```

```python
# 포인트의 평균을 구하고, 리뷰의 포인트값이, 평균보다 큰 데이터 (즉, 평가가 좋은 와인) 만 가져오시오.
df['points'].mean()
```

```python
df[df['points']>df['points'].mean()]
```

```python
# points 컬럼의 median 값은?
df['points'].median()
```

```python
# 각 국가별로는 몇개의 리뷰가 있는지, 각국가별 리뷰수를 구하시오.
df['country'].value_counts()
```

```python
# 나는 경제적이므로, 가격대비 포인트가 가장 큰 와인을 사려한다. 해당 와인의 title은?
```

```python
df['cost-effect'] = df['points'] /df['price']
df[df['cost-effect'] == df['cost-effect'].max()]
```

```python
# 가성비 가장 높은 와인의 title, 나라, 가격까지 가져오기
df.loc[df['cost-effect'] == df['cost-effect'].max(), ['country', 'price']]
```

- 문제 하나를 해결하기 위해서는 하나의 방법으로 풀수 있는게 아니다. 배운것이 쌓여야만, 큰 문제를 해결할 수 있다.
- 작은 문제를 많이 풀어야 큰 문제를 풀 수 있다.
